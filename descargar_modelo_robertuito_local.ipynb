```python
import os
from transformers import AutoTokenizer, AutoModelForSequenceClassification
```

```python
model_checkpoint = "pysentimiento/robertuito-sentiment-analysis"
ruta_destino = "modelo_original/"

os.makedirs(ruta_destino, exist_ok=True)

print(f"Iniciando la descarga de '{model_checkpoint}' desde Hugging Face...")

tokenizer = AutoTokenizer.from_pretrained(model_checkpoint)
tokenizer.save_pretrained(ruta_destino)
print("Tokenizador guardado con éxito.")

model = AutoModelForSequenceClassification.from_pretrained(model_checkpoint)
model.save_pretrained(ruta_destino)
print("Pesos del modelo guardados con éxito.")
```
